# Build the model

```python
#1. 挂载 Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#2. 解压 ZIP 文件
import zipfile
import os

# 指定数据集路径（修改为你的实际路径）
zip_path = "/content/drive/My Drive/Insect Detection.v1i.yolov11.zip"
extract_path = "/content/dataset"

# 解压 ZIP 文件
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("解压完成，文件列表：", os.listdir(extract_path))
```

```python
#3. 确保数据格式正确
import yaml

yaml_path = "/content/dataset/data.yaml"

with open(yaml_path, 'r') as file:
    data_config = yaml.safe_load(file)

print(data_config)
```

```python
#4. 训练 YOLOv11
from ultralytics import YOLO

# 加载 YOLOv11 预训练模型
model = YOLO("yolo11n.pt")  # 你可以选择 yolov11s.pt, yolov11m.pt

# 训练模型
model.train(data="/content/dataset/data.yaml", epochs=50, batch=8, imgsz=640)
```

# Accuracy metrics

```python
#评估模型
metrics = model.val()
```

```python
#查看具体指标
print(f"Precision: {metrics.box.map50:.4f}")  # mAP@50 可能对应 Precision
print(f"Recall: {metrics.box.map75:.4f}")    # mAP@75 可能对应 Recall
print(f"mAP@50: {metrics.box.map:.4f}")      # mAP@50-95
```
